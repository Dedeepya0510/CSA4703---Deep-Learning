```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
customer_genre = np.array([
    [5, 2, 0, 1, 4],
    [4, 1, 1, 0, 5],
    [0, 5, 4, 2, 0],
    [1, 4, 5, 3, 1],
    [3, 0, 1, 5, 4],
    [5, 1, 0, 2, 3]
])
genre_book = np.array([
    [0.9, 0.1, 0.2, 0.7],
    [0.1, 0.9, 0.2, 0.1],
    [0.2, 0.3, 1.0, 0.1],
    [0.1, 0.2, 0.3, 1.0],
    [0.8, 0.1, 0.1, 0.5]
])
print("Customer-Genre Matrix:\n", customer_genre)
print("\nGenre-Book Matrix:\n", genre_book)
print("\nDot Product:",
      np.dot(customer_genre[0], customer_genre[1]))
score = np.dot(customer_genre, genre_book)
print("\nCustomer-Book Score Matrix:\n", score)
print("\nTranspose:\n", customer_genre.T)
A = np.array([[4, 1, 0],
              [1, 3, 1],
              [0, 1, 2]])
print("\nInverse:\n", np.linalg.inv(A))
print("\nAddition:\n", customer_genre + 2)
print("\nMultiplication:\n", customer_genre * 2)
print("\nSquare:\n", customer_genre ** 2)
norm = np.linalg.norm(customer_genre, axis=1)
similarity = (customer_genre @ customer_genre.T) / np.outer(norm, norm)
np.fill_diagonal(similarity, -1)
i, j = np.unravel_index(np.argmax(similarity), similarity.shape)
print("\nCosine Similarity Matrix:\n", similarity)
print("Most Similar Customers:", i+1, "and", j+1)
print("Similarity Score:", similarity[i, j])
x = customer_genre[0]
W = np.array([
    [0.2, 0.1, 0.3, 0.4, 0.2],
    [0.1, 0.3, 0.2, 0.1, 0.4],
    [0.4, 0.2, 0.1, 0.3, 0.1]
])
b = np.array([0.5, 0.2, 0.1])
y = np.dot(W, x) + b
print("\nDense Layer Output:\n", y)
plt.figure(figsize=(8, 5))
sns.heatmap(score, annot=True, cmap="YlGnBu")
plt.title("Customer-Book Recommendation Scores")
plt.xlabel("Books")
plt.ylabel("Customers")
plt.show()
```
