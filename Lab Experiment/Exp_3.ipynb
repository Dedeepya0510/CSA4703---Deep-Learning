```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson, uniform, norm
np.random.seed(42)
data = bernoulli.rvs(0.3, size=1000)
plt.hist(data, bins=2, density=True)
plt.title("Bernoulli Distribution")
plt.show()
data = binom.rvs(10, 0.4, size=1000)
plt.hist(data, bins=15, density=True)
plt.title("Binomial Distribution")
plt.show()
data = poisson.rvs(4, size=1000)
plt.hist(data, bins=15, density=True)
plt.title("Poisson Distribution")
plt.show()
data = uniform.rvs(0, 1, size=1000)
plt.hist(data, bins=20, density=True)
x = np.linspace(0,1,100)
plt.plot(x, uniform.pdf(x,0,1), 'r')
plt.title("Uniform Distribution")
plt.show()
data = norm.rvs(0,1,size=1000)
plt.hist(data, bins=30, density=True)
x = np.linspace(-4,4,100)
plt.plot(x,norm.pdf(x), 'r')
plt.title("Gaussian Distribution")
plt.show()
prevalence = 0.01
sensitivity = 0.95
specificity = 0.90
positive = sensitivity * prevalence + (1-specificity)*(1-prevalence)
posterior = (sensitivity*prevalence)/positive
print("Probability of Disease Given Positive Test:")
print(posterior*100, "%")
N = 100000
disease = np.random.random(N) < prevalence
test_positive = np.where(
    disease,
    np.random.random(N) < sensitivity,
    np.random.random(N) < (1-specificity)
)
empirical = disease[test_positive].mean()
p = np.linspace(0.001,0.20,200)
posterior_values = (
    sensitivity*p /
    (sensitivity*p + (1-specificity)*(1-p))
)
plt.plot(p*100,posterior_values*100)
plt.xlabel("Disease Prevalence (%)")
plt.ylabel("Posterior Probability (%)")
plt.title("Bayesian Posterior vs Prevalence")
plt.grid()
plt.show()
```
